```python
filename = './sample.txt'
with open(filename) as f_read:
    content = f_read.read()
print(content)
```

```python
with open(filename) as f_read:
    lines = f_read.readlines()
    print(lines)

for line in lines:
    print(line.strip())
```

```python
with open(filename) as f_read:
    lines = f_read.readlines()
    print(lines)
```

```python
with open(filename) as f_read:
    while True:
        line = f_read.readline()
        if line:
            print(line)
        else:
            break
```

```python
fruits = ['사과', '토마토', '바나나']
outfile = 'fruits.txt'

with open(outfile, mode='w', encoding='utf-8') as f_write:
    # for fruit in fruits:
    #     f_write.write(fruit + '\n')
    f_write.writelines('\n'.join(fruits))
```

```python
with open(outfile, mode='a', encoding='utf-8') as f_write:
    f_write.write('\n'.join(fruits))
```
